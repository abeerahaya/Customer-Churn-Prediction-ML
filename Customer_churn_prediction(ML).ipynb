```python
# Customer Churn Prediction using Machine Learning
# Data Loading & Exploratory Data Analysis (EDA)

# Import Required Libraries

import warnings
warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd

import matplotlib.pyplot as plt
import seaborn as sns

from IPython.display import display

# Display Settings
pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)

print("="*60)
print("Customer Churn Prediction Project")
print("="*60)

# Load Dataset

from google.colab import drive
drive.mount('/content/drive')

df = pd.read_csv("/content/drive/MyDrive/WA_Fn-UseC_-Telco-Customer-Churn.csv")

print("\nDataset Loaded Successfully!")
print("Shape of Dataset:", df.shape)

# Preview Dataset

print("\nFirst 5 Rows")
display(df.head())

print("\nLast 5 Rows")
display(df.tail())

# Dataset Information

print("\nDataset Information")
df.info()

# Statistical Summary

print("\nNumerical Summary")
display(df.describe())

print("\nCategorical Summary")
display(df.describe(include='object'))

# Check Missing Values

print("\nMissing Values")
missing = df.isnull().sum()

missing_df = pd.DataFrame({
    "Missing Values": missing,
    "Percentage": round((missing/len(df))*100,2)
})

display(missing_df[missing_df["Missing Values"]>0])

# Check Duplicate Records

duplicates = df.duplicated().sum()

print("\nDuplicate Records:", duplicates)

# Dataset Columns

print("\nColumns")

for i,col in enumerate(df.columns,1):
    print(f"{i}. {col}")

# Target Variable Distribution

plt.figure(figsize=(6,4))

sns.countplot(
    x="Churn",
    data=df
)

plt.title("Customer Churn Distribution")
plt.xlabel("Churn")
plt.ylabel("Count")

plt.show()

# Percentage

print("\nTarget Percentage")

display(
    (
        df["Churn"]
        .value_counts(normalize=True)
        *100
    ).round(2)
)

# Data Types

print("\nData Types")

display(df.dtypes)

# Unique Values

print("\nUnique Values in Each Column")

unique_df = pd.DataFrame({
    "Column":df.columns,
    "Unique Values":[df[col].nunique() for col in df.columns]
})

display(unique_df)

# Basic EDA

print("\nNumerical Columns")

num_cols = df.select_dtypes(include=np.number).columns.tolist()

print(num_cols)

print("\nCategorical Columns")

cat_cols = df.select_dtypes(include="object").columns.tolist()

print(cat_cols)

# Correlation (Numerical Only)

if len(num_cols) > 1:

    plt.figure(figsize=(8,6))

    sns.heatmap(
        df[num_cols].corr(),
        annot=True,
        cmap="coolwarm",
        fmt=".2f"
    )

    plt.title("Correlation Matrix")

    plt.show()
```

```python
# Data Cleaning & Feature Engineering

print("=" * 60)
print("Part 2 - Data Cleaning & Feature Engineering")
print("=" * 60)

# 1. Create Backup

df = df.copy()

# 2. Drop Customer ID

if "customerID" in df.columns:
    df.drop("customerID", axis=1, inplace=True)

print("customerID column removed.")

# 3. Convert TotalCharges to Numeric

df["TotalCharges"] = pd.to_numeric(df["TotalCharges"], errors="coerce")

# 4. Missing Values

print("\nMissing Values Before Cleaning:")
print(df.isnull().sum()[df.isnull().sum() > 0])

# Fill missing TotalCharges with Median
df["TotalCharges"].fillna(df["TotalCharges"].median(), inplace=True)

print("\nMissing Values After Cleaning:")
print(df.isnull().sum().sum())

# 5. Remove Duplicate Rows

duplicates = df.duplicated().sum()

print(f"\nDuplicate Rows Before: {duplicates}")

if duplicates > 0:
    df.drop_duplicates(inplace=True)

print(f"Dataset Shape After Removing Duplicates: {df.shape}")

# 6. Binary Encoding

binary_columns = [
    "gender",
    "Partner",
    "Dependents",
    "PhoneService",
    "PaperlessBilling",
    "Churn"
]

binary_map = {
    "Yes":1,
    "No":0,
    "Male":1,
    "Female":0
}

for col in binary_columns:
    df[col] = df[col].map(binary_map)

# 7. Multi-Class Encoding

multi_columns = [
    "MultipleLines",
    "InternetService",
    "OnlineSecurity",
    "OnlineBackup",
    "DeviceProtection",
    "TechSupport",
    "StreamingTV",
    "StreamingMovies",
    "Contract",
    "PaymentMethod"
]

df = pd.get_dummies(
    df,
    columns=multi_columns,
    drop_first=True,
    dtype=int
)

print("\nCategorical Encoding Completed.")

# 8. Separate Features & Target

X = df.drop("Churn", axis=1)

y = df["Churn"]

print("\nFeature Matrix Shape:", X.shape)
print("Target Shape:", y.shape)

# 9. Display Processed Dataset

print("\nProcessed Dataset Preview")

display(X.head())

# 10. Target Distribution

print("\nTarget Distribution")

print(y.value_counts())

print("\nTarget Percentage")

print(round(y.value_counts(normalize=True) * 100,2))

# 11. Final Check

print("\nAny Missing Values Left?")

print(X.isnull().sum().sum())

print("\nFinal Dataset Shape")

print(df.shape)

print("\nData Cleaning Completed Successfully!")
```

```python
# Feature Scaling & Train-Test Split

# Import Required Libraries

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

# Target Distribution

print("\nTarget Distribution")

target_counts = y.value_counts()

print(target_counts)

print("\nTarget Percentage")

print(round(y.value_counts(normalize=True)*100,2))

# Numerical Features

numerical_features = [
    "tenure",
    "MonthlyCharges",
    "TotalCharges"
]

print("\nNumerical Features")
print(numerical_features)

# Train-Test Split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("\nTrain Shape :", X_train.shape)
print("Test Shape  :", X_test.shape)

# Feature Scaling

scaler = StandardScaler()

X_train_scaled = X_train.copy()
X_test_scaled = X_test.copy()

X_train_scaled[numerical_features] = scaler.fit_transform(
    X_train[numerical_features]
)

X_test_scaled[numerical_features] = scaler.transform(
    X_test[numerical_features]
)

print("\nFeature Scaling Completed.")

# Final Verification

print("\nTraining Target Distribution")

print(y_train.value_counts())

print("\nTesting Target Distribution")

print(y_test.value_counts())

# Dataset Summary

summary = pd.DataFrame({

    "Dataset":[
        "Training Features",
        "Testing Features",
        "Training Target",
        "Testing Target"
    ],

    "Rows":[
        X_train_scaled.shape[0],
        X_test_scaled.shape[0],
        y_train.shape[0],
        y_test.shape[0]
    ],

    "Columns":[
        X_train_scaled.shape[1],
        X_test_scaled.shape[1],
        1,
        1
    ]

})

display(summary)

# Preview Scaled Data

print("\nScaled Training Data")

display(X_train_scaled.head())

print("\nScaled Testing Data")

display(X_test_scaled.head())
```

```python
# Model Building & Performance Comparison

from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import (
    RandomForestClassifier,
    ExtraTreesClassifier,
    GradientBoostingClassifier
)

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    classification_report
)

# Optional XGBoost

try:
    from xgboost import XGBClassifier
    xgb_available = True
except ImportError:
    xgb_available = False

# Models

models = {
    "Logistic Regression": LogisticRegression(
        max_iter=1000,
        class_weight="balanced",
        random_state=42
    ),

    "Random Forest": RandomForestClassifier(
        n_estimators=300,
        class_weight="balanced",
        random_state=42,
        n_jobs=-1
    ),

    "Extra Trees": ExtraTreesClassifier(
        n_estimators=300,
        class_weight="balanced",
        random_state=42,
        n_jobs=-1
    ),

    "Gradient Boosting": GradientBoostingClassifier(
        random_state=42
    )
}

if xgb_available:
    models["XGBoost"] = XGBClassifier(
        n_estimators=300,
        learning_rate=0.05,
        max_depth=5,
        subsample=0.8,
        colsample_bytree=0.8,
        eval_metric="logloss",
        random_state=42
    )

# Train Models

results = []
trained_models = {}

for name, model in models.items():

    print("=" * 70)
    print(f"Training: {name}")
    print("=" * 70)

    # Scaling only for Logistic Regression
    if name == "Logistic Regression":
        Xtr = X_train_scaled
        Xte = X_test_scaled
    else:
        Xtr = X_train
        Xte = X_test

    model.fit(Xtr, y_train)

    pred = model.predict(Xte)
    prob = model.predict_proba(Xte)[:, 1]

    trained_models[name] = model

    accuracy = accuracy_score(y_test, pred)
    precision = precision_score(y_test, pred)
    recall = recall_score(y_test, pred)
    f1 = f1_score(y_test, pred)
    auc = roc_auc_score(y_test, prob)

    results.append({
        "Model": name,
        "Accuracy": accuracy,
        "Precision": precision,
        "Recall": recall,
        "F1 Score": f1,
        "ROC-AUC": auc
    })

    print(classification_report(y_test, pred))

# Results Comparison

results_df = pd.DataFrame(results)

results_df = results_df.sort_values(
    by=["ROC-AUC", "F1 Score"],
    ascending=False
).reset_index(drop=True)

print("\nModel Comparison")
display(results_df)

# Best Baseline Model

best_model_name = results_df.loc[0, "Model"]
best_model = trained_models[best_model_name]

print(f"\nBest Baseline Model: {best_model_name}")
```

```python
# Hyperparameter Tuning & Final Model Selection

from sklearn.model_selection import RandomizedSearchCV

print("="*70)
print("Hyperparameter Tuning")
print("="*70)

# Random Forest Parameters

rf_params = {

    "n_estimators":[200,300,500],

    "max_depth":[None,10,20,30],

    "min_samples_split":[2,5,10],

    "min_samples_leaf":[1,2,4],

    "max_features":["sqrt","log2"],

    "bootstrap":[True,False]

}

# Extra Trees Parameters

et_params = {

    "n_estimators":[200,300,500],

    "max_depth":[None,10,20,30],

    "min_samples_split":[2,5,10],

    "min_samples_leaf":[1,2,4],

    "max_features":["sqrt","log2"]

}

# Random Forest Search

rf_search = RandomizedSearchCV(

    RandomForestClassifier(

        random_state=42,

        class_weight="balanced"

    ),

    rf_params,

    n_iter=20,

    scoring="roc_auc",

    cv=5,

    random_state=42,

    verbose=1,

    n_jobs=-1

)

rf_search.fit(X_train,y_train)

# Extra Trees Search

et_search = RandomizedSearchCV(

    ExtraTreesClassifier(

        random_state=42,

        class_weight="balanced"

    ),

    et_params,

    n_iter=20,

    scoring="roc_auc",

    cv=5,

    random_state=42,

    verbose=1,

    n_jobs=-1

)

et_search.fit(X_train,y_train)

# Evaluate Tuned Models

tuned_results=[]

tuned_models={

    "Random Forest Tuned":rf_search.best_estimator_,

    "Extra Trees Tuned":et_search.best_estimator_

}

for name,model in tuned_models.items():

    pred=model.predict(X_test)

    prob=model.predict_proba(X_test)[:,1]

    tuned_results.append({

        "Model":name,

        "Accuracy":accuracy_score(y_test,pred),

        "Precision":precision_score(y_test,pred),

        "Recall":recall_score(y_test,pred),

        "F1 Score":f1_score(y_test,pred),

        "ROC-AUC":roc_auc_score(y_test,prob)

    })

results_tuned=pd.DataFrame(tuned_results)

display(results_tuned)

# Combine Baseline + Tuned Models

all_results=pd.concat(

    [results_df,results_tuned],

    ignore_index=True

)

all_results=all_results.sort_values(

    by=["ROC-AUC","F1 Score"],

    ascending=False

).reset_index(drop=True)

print("="*70)
print("Overall Model Comparison")
print("="*70)

display(all_results)

# Select Final Best Model

best_model_name=all_results.iloc[0]["Model"]

if best_model_name=="Random Forest Tuned":

    best_model=rf_search.best_estimator_

elif best_model_name=="Extra Trees Tuned":

    best_model=et_search.best_estimator_

else:

    best_model=trained_models[best_model_name]

print("="*70)
print("FINAL BEST MODEL")
print(best_model_name)
print("="*70)

# Best Parameters

if best_model_name=="Random Forest Tuned":

    print(rf_search.best_params_)

elif best_model_name=="Extra Trees Tuned":

    print(et_search.best_params_)
```

```python
# Final Evaluation

import joblib
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.metrics import (
    confusion_matrix,
    ConfusionMatrixDisplay,
    RocCurveDisplay,
    PrecisionRecallDisplay,
    classification_report
)

from sklearn.model_selection import cross_val_score

print("="*70)
print("FINAL MODEL EVALUATION")
print("="*70)

# Predictions

pred = best_model.predict(X_test)
prob = best_model.predict_proba(X_test)[:,1]

# Classification Report

print("\nClassification Report\n")

print(classification_report(y_test,pred))

# Confusion Matrix

cm = confusion_matrix(y_test,pred)

plt.figure(figsize=(6,5))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=["No Churn","Churn"],
    yticklabels=["No Churn","Churn"]
)

plt.title("Confusion Matrix")
plt.xlabel("Predicted")
plt.ylabel("Actual")

plt.show()

# ROC Curve

RocCurveDisplay.from_estimator(
    best_model,
    X_test,
    y_test
)

plt.title("ROC Curve")

plt.show()

# Precision Recall Curve

PrecisionRecallDisplay.from_estimator(
    best_model,
    X_test,
    y_test
)

plt.title("Precision Recall Curve")

plt.show()

# Cross Validation

cv_scores = cross_val_score(

    best_model,

    X,

    y,

    cv=5,

    scoring="roc_auc",

    n_jobs=-1

)

print("="*60)

print("Cross Validation ROC-AUC Scores")

print(cv_scores)

print("\nAverage ROC-AUC")

print(round(cv_scores.mean(),4))

print("="*60)

# Feature Importance

if hasattr(best_model,"feature_importances_"):

    importance = pd.DataFrame({

        "Feature":X.columns,

        "Importance":best_model.feature_importances_

    })

    importance = importance.sort_values(

        by="Importance",

        ascending=False

    )

    display(importance.head(20))

    plt.figure(figsize=(10,8))

    sns.barplot(

        data=importance.head(15),

        x="Importance",

        y="Feature"

    )

    plt.title("Top 15 Important Features")

    plt.show()

# Save Model

joblib.dump(best_model,"customer_churn_model.pkl")

print("\nModel Saved Successfully!")

# Prediction Function

def predict_customer(sample):

    prediction = best_model.predict(sample)

    probability = best_model.predict_proba(sample)

    if prediction[0]==1:

        print("Customer is likely to CHURN")

    else:

        print("Customer is likely to STAY")

    print("\nProbability")

    print(probability)

# Project Summary

print("Customer Churn Prediction Project Completed Successfully")

print("Best Model :",best_model_name)
```
